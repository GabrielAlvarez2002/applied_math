```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

item 1

```python
x = sp.symbols("X")
def calcular_integral_n(n):
  f_n = (x**n) * sp.exp(x)
  resultado = sp.integrate(f_n, (x, 0, 1))
  res = resultado
  return res

```

```python
print(calcular_integral_n(0))
```

item 3

```python
vals_n = list(range(1,26))
resultados_integrales = []
for n_val in vals_n:
  resultados_integrales.append(calcular_integral_n((n_val)))
print(resultados_integrales)
```

```python
# Convierto los resultados simbolicos en valores numerables
resultados_numericos = [sp.N(res) for res in resultados_integrales]
plt.figure(figsize=(10, 6))
plt.plot(vals_n, resultados_numericos)
plt.title("Valor de la integral")
plt.xlabel("n (valores de prueba)")
plt.ylabel("Valor")
plt.grid(True)
plt.show()
```

item 1

```python
def sucesion_collatz(n):
    a = [n]
    while n > 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        a.append(n)
    return a
```

```python
print(sucesion_collatz(27))
```

item 2

```python
def pasos_collatz(n):
    contador = 0
    while n > 1:
        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1
        contador += 1
    return contador
print(pasos_collatz(27))
```

```python
def max_collatz(n):
  max_val = n
  while n>1:
    if n % 2 == 0:
      n //= 2
    else:
      n = 3 * n + 1
    if n > max_val:
      max_val = n
  return max_val
print(max_collatz(27))
```

```python
secuencia_27 = sucesion_collatz(27)
max_val_27 = max_collatz(27)
max_val_index = secuencia_27.index(max_val_27) # esto me dice en que paso de la secuencia se encuentra el val maximo

plt.figure(figsize=(12, 7))
plt.plot(range(len(secuencia_27)), secuencia_27,"*" , label='secuencia de para n=27')
plt.plot(max_val_index, max_val_27, "ro", markersize=10, label='valor maximo')
plt.title('secuencia de para n=27')
plt.xlabel('numero de pasos')
plt.ylabel('Valor')
plt.grid(True)
plt.legend()
plt.show()
```

item 3

```python
n = np.arange(1,10000)
for i in n:
    print("cant_pasos =",pasos_collatz(i))
```

```python
n_valores = np.arange(1, 10001)
lista = []
for val in n_valores:
  lista.append(pasos_collatz(int(val)))

plt.figure(figsize=(12, 6))
plt.plot(n_valores, lista)
plt.title('Cantidad de pasos de Collatz en función de n')
plt.xlabel('Valor de n')
plt.ylabel('Cantidad de pasos')
plt.grid(True)
plt.show()
```

```python
def busqueda_binaria(lista, v):
  left = 0
  right = len(lista) - 1
  while left <= right:
    mid = (left + right) // 2
    if lista[mid] == v:
      return mid
    elif lista[mid] < v:
      left = mid + 1
    else:
      right = mid - 1
  return -1
```

```python
print(busqueda_binaria([2, 5, 8, 12, 16, 23, 38, 42, 77, 91],11))
print(busqueda_binaria([2, 5, 8, 12, 16, 23, 38, 42, 77, 91],12))
```
