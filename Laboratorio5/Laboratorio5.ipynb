```python
def solucion_exacta(x, e_val):
  return 1+x+((np.exp(x/e_val)-1)/(np.exp(1/e_val)-1))
```

```python
x = np.linspace(0,1,1000)
e = [0.2,0.05,0.01,0.002]
```

```python
for e_val in e:
  plt.plot(x,solucion_exacta(x, e_val))
plt.grid()
plt.show()
```

```python
import numpy as np

def matriz(N_dim, e_val):
  h_val = 1.0 / (N_dim + 1)

  a_coeff = (e_val / h_val**2) + (1.0 / (2 * h_val))
  b_coeff = (-2.0 * e_val) / h_val**2
  c_coeff = (e_val / h_val**2) - (1.0 / (2 * h_val))


  main_diag_elements = b_coeff * np.ones(N_dim)
  super_diag_elements = a_coeff * np.ones(N_dim - 1)
  sub_diag_elements = c_coeff * np.ones(N_dim - 1)

  M = np.diag(main_diag_elements)

  if N_dim > 1:
      M += np.diag(super_diag_elements, k=1)
      M += np.diag(sub_diag_elements, k=-1)

  return M
```
